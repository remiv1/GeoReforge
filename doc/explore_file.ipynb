## Catalogue Natural Earth

Les couches et leurs précisions sont chargées depuis `naturalearth.json`. Le filtre par famille et la recherche permettent de retrouver rapidement un fichier.

## Sélection interactive

Choisissez un fichier dans la première colonne, cochez une ou plusieurs précisions disponibles dans la deuxième, puis ajoutez-le. La troisième colonne récapitule les couches sélectionnées et permet de les retirer.

### Chargement

```python
from pathlib import Path    # pylint: disable=C0114
import sys

from IPython.display import display # type: ignore

NE_FILE = "naturalearth.json"
current_dir = Path.cwd()
if (current_dir / NE_FILE).is_file():
    doc_dir = current_dir
elif (current_dir / "doc" / NE_FILE).is_file():
    doc_dir = current_dir / "doc"
else:
    raise FileNotFoundError(f"{NE_FILE} est introuvable depuis le répertoire courant.")

PROJECT_DIR = doc_dir.parent
SCRIPTS_PATH = str(PROJECT_DIR / "scripts")
if SCRIPTS_PATH not in sys.path:
    sys.path.insert(0, SCRIPTS_PATH)

from zones_maker.notebook_ui import ZonesNotebook   # type: ignore  # pylint: disable=E0401, C0413

notebook_app = ZonesNotebook(
    doc_dir / NE_FILE,
    PROJECT_DIR / "tmp",
    PROJECT_DIR / "src" / "zones.toml",
)
print(f"Catalogue chargé : {len(notebook_app.datasets)} couches.")
```

### Résumé

La colonne de droite affiche les fichiers ajoutés, leur famille et les précisions retenues. Le bouton de retrait supprime une couche de la sélection.
