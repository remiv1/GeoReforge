# Selection and exploration of Natural Earth datas

_Sélection et exploration des données Natural Earth_

This notebook allows you to browse the Natural Earth catalog and select the available files along with their details.

_Ce notebook permet de parcourir le catalogue Natural Earth et de sélectionner les fichiers ainsi que leurs précisions disponibles._

## Natural Earth Catalog

_Catalogue Natural Earth_

The layers and their details are loaded from `naturalearth.json`. The family filter and search allow you to quickly find a file.

_Les couches et leurs précisions sont chargées depuis `naturalearth.json`. Le filtre par famille et la recherche permettent de retrouver rapidement un fichier._

## Interactive Selection

_Sélection interactive_

Select a file in the first column, check one or more available details in the second column, and then add it. The third column summarizes the selected layers and allows you to remove them.

_Choisissez un fichier dans la première colonne, cochez une ou plusieurs précisions disponibles dans la deuxième, puis ajoutez-le. La troisième colonne récapitule les couches sélectionnées et permet de les retirer._

### Loading

_Chargement_

Launch the cell to load the Natural Earth catalog.

_Lancez la cellule pour charger le catalogue Natural Earth._

```python
from pathlib import Path    # pylint: disable=C0114
import sys

from IPython.display import display # type: ignore

PROJECT_DIR = Path.cwd().parent
NE_FILE = "naturalearth.json"
CATALOG_PATH = PROJECT_DIR / "src" / NE_FILE
if not CATALOG_PATH.is_file():
    raise FileNotFoundError(f"{CATALOG_PATH} est introuvable.")

SCRIPTS_PATH = str(PROJECT_DIR / "scripts")
if SCRIPTS_PATH not in sys.path:
    sys.path.insert(0, SCRIPTS_PATH)

from zones_maker.notebook_ui import ZonesNotebook   # type: ignore  # pylint: disable=E0401, C0413

notebook_app = ZonesNotebook(
    CATALOG_PATH,
    PROJECT_DIR / "tmp",
    PROJECT_DIR / "src" / "zones.toml",
)
print(f"Catalogue chargé : {len(notebook_app.datasets)} couches.")
```

### Summary

_Résumé_

The right column displays the added files, their family, and the selected details. The remove button deletes a layer from the selection.

_La colonne de droite affiche les fichiers ajoutés, leur famille et les précisions retenues. Le bouton de retrait supprime une couche de la sélection._

```python
display(notebook_app.selection_view())
```

_Exploration et mapping_

_Actualisez la liste après vos sélections, puis choisissez un fichier et une précision. Le shapefile n’est téléchargé dans `tmp/` que s’il manque et que vous lancez son exploration. Les correspondances de colonnes et de classes restent à votre choix._
